# Task 2 : Categories(Sets)

```python
# Setup data based on the Task 1 instructions ( Paralles list categories)

products = ['Laptop', 'Smartphone', 'Smartwatch', 'Laptop', 'iPad', 'i Watch', 'Telivision', 'Smart Watch']

categories = ['Electronics','Electronics','Accessories','Electronics','Electronics','Accessories','Electronics','Accessories']

# 1. From the product list , create a set of categories called categories_set

categories_set = set(categories)
print("categories_set :", categories_set)
```

```python
# Demonstrate adding a new category to the set and show that duplicate are ignored.

categories_set.add("Gaming")
categories_set.add("Electronics")
print('after trying to add Gaming and duplicate Electronics : ', categories_set)
```

```python
# Show how to check whether a category exist in the set( print a boolean results)

check_categories = "Gaming"
exists = check_categories in categories_set
print(f'Dose {check_categories} exist in the set? ', exists)
```

```python
# Extra (Optional): - show how to get the total number of unique categories using set.

total_unique = len(categories_set)
print('Total number of unique categories ', total_unique)
```

# Task 3 Product Pricing (Dictionaries)

```python
# Create a dictionary price_dict where keys are product name and value are price (integer or floats). Include at least 6 entries.

price_dict = {'Laptop': 120000,
              'smartphone':80000,
              'smartwatch':20000,
              'iPad':89000,
              'Telivision':56000,
              'Monitor':46000}
```

```python
# write small code blocks to:
#Add a new procuct with price to price_dict.

price_dict['earbuds'] = 2100
print('After adding the earbuds', price_dict)
```

```python
# update the price of an existing products
product_to_update = 'smartwatch'

if product_to_update in price_dict:
    price_dict[product_to_update]=26000
print(f'Updated {product_to_update} price',price_dict[product_to_update])
```

```python
# Remove a product by name (handle the case when the product does not exist)

product_to_remove = 'earphone'

if product_to_remove in price_dict:
   del price_dict[product_to_remove]
   print(f'Removed{product_to_remove}.')
else:
   print(f'Cannot removed {product_to_remove}: product does not exist')
```

```python
# print the avarage price of all products (use only dictionary operations and basic arithmatic)

total_price = sum(price_dict.values())
num_products = len(price_dict)
average_price = total_price / num_products
print(f'Average price of all products : {average_price}')
```

```python
# print the avarage price of all products (use only dictionary operations and basic arithmatic)

total_price = sum(price_dict.values())
num_products = len(price_dict)
average_price = total_price / num_products
print(f'Average price of all products : {average_price:.2f}')
```

```python
# print the avarage price of all products (use only dictionary operations and basic arithmatic)

total_price = sum(price_dict.values())
num_products = len(price_dict)
average_price = total_price / num_products
print(f'Average price of all products : {average_price:.3f}')
```

```python
# Extra ( Optional): Print the product with both the maximum and minimum price

max_product = max(price_dict,key=price_dict.get)
min_product = min(price_dict,key=price_dict.get)

print(f'Most expensive product : {max_product} ({price_dict[max_product]:.2f})')
print(f'Least expensive product : {min_product}({price_dict[min_product]:.2f})')
```
